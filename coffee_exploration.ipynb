Loading CSV Files

```python
import pandas as pd

exports_df = pd.read_csv("Coffee_export.csv")
imports_df = pd.read_csv("Coffee_import.csv")
production_df = pd.read_csv("Coffee_production.csv")

print("Coffee Exports")
display(exports_df.head())

print("Coffee Imports")
display(imports_df.head())

print("Coffee Production")
display(production_df.head())
```

Explore the Dataset

```python
def explore_dataset(df, name):
    print(f"\n{name}")
    print("-" * 40)

    print("Shape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())
```

Run the Exploration

```python
explore_dataset(exports_df, "Coffee Exports")
explore_dataset(imports_df, "Coffee Imports")
explore_dataset(production_df, "Coffee Production") 
```

Add Years Represented and Countries Represented

```python
print("Export Countries:")
print(exports_df["Country"].unique())

print("\nNumber of Export Countries:")
print(exports_df["Country"].nunique())

print("\nImport Countries:")
print(imports_df["Country"].unique())

print("\nNumber of Import Countries:")
print(imports_df["Country"].nunique())

print("\nProduction Countries:")
print(production_df["Country"].unique())

print("\nNumber of Production Countries:")
print(production_df["Country"].nunique())
```

## Dataset Exploration Findings

### Coffee Exports

- Contains 55 rows and 32 columns.
- Includes export data from 1990 through 2019.
- All columns contain valid data with no missing values.
- Country is stored as text and all export values are stored as integers.

### Coffee Imports

- Contains 35 rows and 32 columns.
- Includes import data from 1990 through 2019.
- No missing values were detected.
- Country is stored as text and all import values are stored as integers.

### Coffee Production

- Contains 55 rows and 33 columns.
- Includes production\-season data from 1990/91 through 2019/20.
- Contains both Country and Coffee Type fields.
- No missing values were detected.
- Production values are stored as floating\-point numbers.

## Statistical Observations

### Coffee Exports

- Coffee export volumes vary substantially across countries and years.
- The highest total export value in the dataset is 33,807,709,056.
- Some countries export significantly more coffee than others, resulting in a large gap between minimum and maximum values.
- Export activity appears concentrated among a relatively small number of major exporting nations.

### Coffee Imports

- Coffee imports are also highly concentrated among a small number of countries.
- The highest total import value in the dataset is approximately 5.07 billion.
- Average import values are considerably lower than the maximum values, indicating that a few countries account for a large percentage of global coffee imports.
- Import trends show substantial variation across the years represented in the dataset.

### Coffee Production

- Coffee production varies greatly among producing countries and coffee types.
- The highest total production value in the dataset is approximately 4.10 billion.
- Production values show significant year\-to\-year fluctuations.
- The wide range between minimum and maximum production values suggests that a small number of countries dominate global coffee production.
